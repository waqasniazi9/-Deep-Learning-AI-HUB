```python
import ipywidgets as widgets
from IPython.display import HTML, display
import os
from pathlib import Path

# Get all notebooks
notebooks = sorted([f for f in os.listdir('.') if f.endswith('.ipynb') and not f.startswith('.')])

print(f"Total Notebooks Found: {len(notebooks)}")
```

```python
# Display notebooks by category
def categorize(filename):
    prefix = filename.split('_')[0]
    if prefix.startswith('1'):
        return 'PyTorch Basics'
    elif prefix.startswith('2'):
        return 'Deep Learning Theory'
    else:
        return 'Advanced Topics'

# Create category breakdown
categories = {}
for nb in notebooks:
    cat = categorize(nb)
    if cat not in categories:
        categories[cat] = []
    categories[cat].append(nb)

# Display
for cat, nbs in sorted(categories.items()):
    print(f"\n📚 {cat}: {len(nbs)} notebooks")
    for nb in nbs[:3]:  # Show first 3
        print(f"  - {nb}")
    if len(nbs) > 3:
        print(f"  ... and {len(nbs) - 3} more")
```

```python
# Interactive notebook selector
dropdown = widgets.Dropdown(
    options=notebooks,
    description='Select Notebook:',
    style={'description_width': 'initial'}
)

def show_info(change):
    nb = change['new']
    size = os.path.getsize(nb) / 1024  # KB
    print(f"\n📓 {nb}")
    print(f"Size: {size:.2f} KB")
    print(f"Category: {categorize(nb)}")

dropdown.observe(show_info, names='value')
display(dropdown)
```
